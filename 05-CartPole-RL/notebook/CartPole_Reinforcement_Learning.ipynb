# CartPole Reinforcement Learning

### Objective

Develop a Reinforcement Learning (RL) agent to balance a pole on a moving cart using the CartPole-v1 environment from Gymnasium.

### Environment

CartPole-v1

### Goal

Train an intelligent agent that learns how to balance the pole for as long as possible by interacting with the environment.

```python
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
```

# Creating the CartPole Environment

Initialize the CartPole-v1 environment provided by Gymnasium.

```python
env = gym.make("CartPole-v1")
```

# Environment Information

Display the observation space and action space of the CartPole environment.

```python
print("Observation Space:", env.observation_space)
print("Action Space:", env.action_space)
```

# Reset the Environment

Reset the environment to obtain the initial state before starting an episode.

```python
state, info = env.reset()

print("Initial State:")
print(state)
```

# Simulate a Random Agent

Execute one episode where the agent selects random actions.

```python
state, info = env.reset()

done = False
total_reward = 0

while not done:
    action = env.action_space.sample()

    state, reward, terminated, truncated, info = env.step(action)

    total_reward += reward

    done = terminated or truncated

print("Total Reward:", total_reward)
```

# Evaluate Random Agent Performance

Run multiple episodes and record the rewards.

```python
rewards = []

for episode in range(20):

    state, info = env.reset()

    done = False
    total_reward = 0

    while not done:

        action = env.action_space.sample()

        state, reward, terminated, truncated, info = env.step(action)

        total_reward += reward

        done = terminated or truncated

    rewards.append(total_reward)

print("Average Reward:", np.mean(rewards))
```

# Visualize Episode Rewards

Plot the total reward obtained in each episode to observe the performance of the agent.

```python
plt.figure(figsize=(8,5))

plt.plot(rewards, marker='o')

plt.title("Reward per Episode")
plt.xlabel("Episode")
plt.ylabel("Total Reward")

plt.grid(True)

plt.show()
```

# Reward Statistics

Calculate the minimum, maximum, and average reward across all episodes.

```python
print("Minimum Reward :", np.min(rewards))
print("Maximum Reward :", np.max(rewards))
print("Average Reward :", np.mean(rewards))
```

# Save Reward Data

Save the episode rewards for future analysis.

```python
np.save("../models/cartpole_rewards.npy", rewards)

print("Reward data saved successfully.")
```

# Environment Details

Display basic information about the CartPole environment.

```python
print("Observation Space:", env.observation_space)
print("Action Space:", env.action_space)
print("Reward Threshold:", env.spec.reward_threshold)
print("Maximum Episode Steps:", env.spec.max_episode_steps)
```

# Close the Environment

Release the resources used by the CartPole environment.

```python
env.close()

print("Environment closed successfully.")
```
