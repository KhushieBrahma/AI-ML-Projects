# Create the LunarLander Environment

The LunarLander environment simulates a spacecraft attempting to land safely on a designated landing pad. The reinforcement learning agent learns through trial and error by interacting with the environment.

```python
env = gym.make("LunarLander-v2", render_mode="rgb_array")
```

# Environment Information

```python
print("Observation Space:", env.observation_space)
print("Action Space:", env.action_space)
```

# Initialize the Deep Q-Network (DQN)

A Deep Q-Network uses a neural network to estimate Q-values for each possible action. The agent learns the optimal policy by maximizing cumulative future rewards.

```python
model = DQN(
    policy="MlpPolicy",
    env=env,
    learning_rate=0.001,
    buffer_size=50000,
    learning_starts=1000,
    batch_size=32,
    gamma=0.99,
    target_update_interval=500,
    verbose=1,
)
```

# DQN Model Initialized Successfully

```python
print(model)
```

## Train the DQN Agent

Train a Deep Q-Network (DQN) agent to learn how to land the Lunar Lander safely.

```python
# Train the model
model.learn(total_timesteps=50000)

print("Training completed successfully!")
```

## Evaluate the Trained Model

```python
mean_reward, std_reward = evaluate_policy(
    model,
    env,
    n_eval_episodes=10
)

print(f"Mean Reward: {mean_reward:.2f}")
print(f"Standard Deviation: {std_reward:.2f}")
```

## Test the Agent

```python
obs, info = env.reset()

done = False
total_reward = 0

while not done:
    action, _ = model.predict(obs, deterministic=True)
    obs, reward, terminated, truncated, info = env.step(action)

    total_reward += reward
    done = terminated or truncated

print("Total Reward:", total_reward)
```

## Save the Trained Model

```python
model.save("../models/lunarlander_dqn")

print("Model saved successfully!")
```
